###  The Objective 
 Using Selenium with Python to scrape real 
 job listings from a live job portal (Naukrigulf) and store the extracted data
 in a structured CSV file.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import time

options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
#options.add_argument("--headless=new")
options.add_argument("--window-size=1500,950")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(options=options)
driver.get("https://www.naukrigulf.com/")

time.sleep(2)
driver.find_element(By.XPATH,"//button[@type='button']").click()
driver.find_element(By.NAME,'qsbKey').send_keys('Data Engineer')
time.sleep(1)
driver.find_element(By.XPATH,"//input[@value='Search Jobs']").click()
time.sleep(3)

list_jobs = []

for i in range(3):
      print('page', i+1)
      jobs = driver.find_elements(By.CLASS_NAME,"srp-tuple")
      for job in jobs:
          try:
             job_title = job.find_element(By.CLASS_NAME,"designation-title").text
             company_name = job.find_element(By.CLASS_NAME,"info-org").text
             job_location = job.find_element(By.CLASS_NAME,"info-loc").text
             experience = job.find_element(By.CLASS_NAME,"info-exp ").text
             description = job.find_element(By.CLASS_NAME,"description").text
             
             list_jobs.append({
                 'job_title':job_title,
                 'company_name':company_name,
                 'job_location':job_location,
                 'experience':experience,
                 'description':description
             })
          except:
              print('Error')
      
      driver.find_element(By.XPATH,"//span[@class='ico bwd']").click()
      time.sleep(2)
```
